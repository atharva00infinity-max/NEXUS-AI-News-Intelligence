now moving to padding and truncation
since all the articles are of different length,we will use padding and truncating so that model gets input of same length

```python
small_train=dataset['train'].select(range(5000)) #creating a small training data set for our pipeline to make sure everything runs correctly
```

```python
small_validation=dataset['validation'].select(range(5000)) 
```

creating a preprocessing function

```python
def preprocess_function(examples):
    model_inputs=tokenizer(  # examples from the article dataset are taken as input, which goes in the t5 tokenizer and  then gets out as token id which will be used by our transformer
        examples['article'],
        max_length=512,
        truncation=True
    )
    labels=tokenizer(
        text_target=examples['highlights'],
        max_length=128,
        truncation=True
        
    )
    model_inputs['labels']=labels['input_ids'] # model_inputs is a dictionary in which we add a key value pair named label that contains input ids
    return model_inputs
```

applying the preprocessing to training data set

```python
tokenized_train=small_train.map(
    preprocess_function,
    batched=True #batched= true means our function recieves multiple examples at once
)
```

```python
tokenized_validation = small_validation.map(
    preprocess_function,
    batched=True
)
```

```python
print(tokenized_train)
```

now loading t5 transformer model

```python
from transformers import AutoModelForSeq2SeqLM
model= AutoModelForSeq2SeqLM.from_pretrained('t5-small')
```

```python
from transformers import DataCollatorForSeq2Seq
# data collector is component that basically prepares individual tokenized examples into a proper batch for t5,
#instead of padding we are using this, it dynamically pads each batch according to the longest example in that batch, hence helping us save a lot of memory
data_collator = DataCollatorForSeq2Seq(
    tokenizer=tokenizer,
    model=model
)
```

```python
from transformers import Seq2SeqTrainingArguments
training_args=Seq2SeqTrainingArguments(
    output_dir='./models/t5-news-summarizer',# where trained model checkpoints will be saved
    num_train_epochs=2,#the model sees your 5,000 training examples twice
    per_device_train_batch_size=4,#processes 4 example at a time
    per_device_eval_batch_size=4,
    learning_rate=2e-5,# how stromgly model updates its weight
    logging_steps=100,
    eval_strategy='epoch',#evaluate after each epoch
    save_strategy='epoch',#save the model after each epoch
    predict_with_generate=True,#lets t5 generate actual summmaries during evaluation
   
)
```

creating a trainer. a trainer connects all pieces together and handles the training loop 

```python
from transformers import Seq2SeqTrainer
trainer=Seq2SeqTrainer(
    model=model, #model is t5
    args=training_args,# how to train
    train_dataset=tokenized_train,#5000 articles + summaries
    eval_dataset=tokenized_validation,#500 validation examples
    data_collator=data_collator,#prepares batches
    processing_class=tokenizer)# handles tokenizing process
```

```python
final_model_path = "./models/t5-news-summarizer-final"

trainer.save_model(final_model_path)
tokenizer.save_pretrained(final_model_path)
```

making our trained model generate a summary

```python
summary_ids=model.generate(
    **inputs,#gives the model the tokenized article (input_ids + attention_mask)
    max_length=128,
    num_beams=4,# generate a summary while considering 4 possible generation paths at a time
    early_stopping=True
)
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_path = "./models/model-35k"

tokenizer = AutoTokenizer.from_pretrained(model_path)
model = AutoModelForSeq2SeqLM.from_pretrained(model_path)

print("Model loaded successfully!")
```

```python
article = small_validation[0]["article"]

inputs = tokenizer(
    article,
    return_tensors="pt",
    max_length=512,
    truncation=True
)

summary_ids = model.generate(
    **inputs,
    max_length=128,
    num_beams=4,
    early_stopping=True
)

generated_summary = tokenizer.decode(
    summary_ids[0],
    skip_special_tokens=True
)

print("Generated Summary:")
print(generated_summary)
```

```python
article = small_validation[0]["article"]
actual_summary = small_validation[0]["highlights"]

inputs = tokenizer(
    article,
    return_tensors="pt",
    max_length=512,
    truncation=True
)

summary_ids = model.generate(
    **inputs,
    max_length=128,
    num_beams=4,
    early_stopping=True
)

generated_summary = tokenizer.decode(
    summary_ids[0],
    skip_special_tokens=True
)

print("ORIGINAL ARTICLE:")
print(article[:1000])

print("\nACTUAL SUMMARY:")
print(actual_summary)

print("\nMODEL-GENERATED SUMMARY:")
print(generated_summary)
```

```python
from rouge_score import rouge_scorer

scorer = rouge_scorer.RougeScorer(
    ["rouge1", "rouge2", "rougeL"],
    use_stemmer=True
)

scores = scorer.score(
    actual_summary,
    generated_summary
)

print("ROUGE Scores:")

for metric, values in scores.items():
    print(f"{metric}: {values.fmeasure:.4f}")
```
